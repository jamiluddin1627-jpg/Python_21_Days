User Authentication System

User Name Validation

```python
import re

u_name_p = r"^[a-z A-Z 0-9_.]{5,15}$"
u_name = input("Enter User name: ")

if re.match(u_name_p, u_name):
    print(u_name)
else:
    print("Invalid User name")
```

User Email Validation

```python
u_mail = input("Email: ")
u_mail_p = r"^\w{3,}+@\w+\.\w{2,}"
if re.match(u_mail_p, u_mail):
    print(u_mail)
else:
    print("Invalid email")
```

User Password Validation

```python
import re

# Strong password regex pattern:
# ^                 : Start of string
# (?=.*[a-z])       : Requires at least one lowercase letter
# (?=.*[A-Z])       : Requires at least one uppercase letter
# (?=.*\d)          : Requires at least one digit
# (?=.*[@$!%*?&#])  : Requires at least one special character
# [A-Za-z\d@$!%*?&#]{8,} : Allowed characters and minimum length of 8
# $                 : End of string
u_pass_p = r"^(?=.*[a-z])(?=.*[A-Z])(?=.*\d)(?=.*[@$!%*?&#])[A-Za-z\d@$!%*?&#]{8,}$"

u_pass = input("Enter Password: ")

# Validate the password against the pattern
if re.match(u_pass_p, u_pass):
    print("Strong password!")
else:
    print(
        "Weak password! Must be at least 8 characters long and include: uppercase, lowercase, number, and special character."
    )
```
