Class Inheritance

```python
# Single Inheritance
# 1 parent ---> 1 child
class Person:
    def __init__(self, name, age, salary): # constructor method
        self.name = name # public attribute can be accessed from anywhere
        self._age = age  # protected attribute can be accessed within the class and its subclasses
        self.__salary = salary # private attribute can only be accessed within the class
    
    # getter method for private attribute
    def get_salary(self):
        return self.__salary

# p1 = Person("Alice", 30, 50000)
# p1.name
# p1.salary # This will raise an AttributeError because __salary is private
# p1.get_salary() # using getter method to access private attribute
# p1._age # accessing protected attribute, but it's not recommended to do so outside the class or subclass

class Employee(Person): # Employee class inherits from Person class
    def __init__(self, name, age, salary, employee_id):
        super().__init__(name, age, salary) # calling the constructor of the parent class
        self.employee_id = employee_id

#------- From Parent Class -------

p1 = Person("Jamil", 25, 50000)
print("--------------Manager-----------------")
print(p1.name) # accessing public attribute
print(p1._age) # accessing protected attribute (not recommended)
print(p1.get_salary()) # accessing private attribute using getter method

#------- From Child Class -------
e1 = Employee("John", 25, 40000, "E001")
print("--------------Employee-----------------")
print(e1.name) # accessing public attribute from parent class
print(e1._age) # accessing protected attribute from parent class (not recommended)
print(e1.get_salary()) # accessing private attribute from parent class
print(e1.employee_id) # accessing attribute from child class
```
