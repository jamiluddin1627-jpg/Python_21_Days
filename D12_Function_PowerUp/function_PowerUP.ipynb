Recursion

```python
def sum_N(n):
    if n == 0:
        return 0
    return n + sum_N(n - 1)

print(f" Sum of numbers from 1 to N is: {sum_N(5)}")

def factorial(n):
    if n == 1:
        return 1
    return n * factorial(n - 1)

print(f" Factorial of 5 is: {factorial(5)}")
```

Lambda Function

```python
def sqr(n):
    return n*n
sqr(5)
```

```python
# lambda function
sqr = lambda n: n*n
sqr(5)
```

Map Function

```python
# map (function, iterable)

nums = [1, 2, 3, 4, 5]
nums_sqr = list(map(lambda n: n*n, nums))
print(nums_sqr)
```

Filter Function

```python
# Filter (function, iterable)
nums = [1, 2, 3, 4, 5]
even_nums = list(filter(lambda n: n % 2 == 0, nums))
print(even_nums)
```

```python
nums = [1, 2, 3, 4, 5]

# Assign the recursive lambda to a name first
fact = lambda n: 1 if n <= 1 else n * fact(n - 1)

factorial_nums = list(map(fact, nums))
print(factorial_nums)  # Output: [1, 2, 6, 24, 120]
```

Reduce Function

```python
# reduce (function, iterable)

from functools import reduce

nums = [1, 2, 3, 4, 5]
sum_nums = reduce(lambda x, y: x + y, nums)
print(sum_nums)  # Output: 15
```

Decorators

```python
def my_decorator(func):
    def wrapper():
        print("Before the function is called.")
        func()
        print("After the function is called.")
    return wrapper

def say_hello():
    print("Hello!")

say_hello = my_decorator(say_hello)
say_hello()
```

```python
def my_decorator(func):
    def wrapper(*args, **kwargs):
        print("Before the function is called.")
        result = func(*args, **kwargs)
        print("After the function is called.")
        return result
    return wrapper

@my_decorator
def say_hello():
    print("Hello!")

@my_decorator
def sum_numbers(a, b):
    return a + b

@my_decorator
def multiply_numbers(a, b):
    return a * b

# --- Output / Results ---

# 1. No return value to capture, just execute it
say_hello()
print()

# 2. Store or directly print the returned values
sum_res = sum_numbers(5, 10)
print(f"Result of sum: {sum_res}\n")

mult_res = multiply_numbers(4, 3)
print(f"Result of multiplication: {mult_res}")
```

Generators

```python
def count_upto(n):
    for i in range(1, n + 1):
        yield i

print(list(count_upto(5)))  # Output: [1, 2, 3, 4, 5]

for x in count_upto(5):
    print(x)
```
