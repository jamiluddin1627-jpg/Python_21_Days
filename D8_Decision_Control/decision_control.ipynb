IF-ELSE Statement

```python
discount = 0
product_price = float(input("Enter the product price: "))
if product_price >= 1000:
    discount = product_price * 10/100
    actual_price = product_price - discount
    print(f"Discount: {discount}")
    print(f"Actual Price: {actual_price}")
elif product_price >= 5000:
    discount = product_price * 20/100
    actual_price = product_price - discount
    print(f"Discount: {discount}")
    print(f"Actual Price: {actual_price}")
elif product_price >= 10000:
    discount = product_price * 30/100
    actual_price = product_price - discount
    print(f"Discount: {discount}")
    print(f"Actual Price: {actual_price}")
else:
    print("No discount applicable.")
    print(f"Actual Price: {product_price}")
```

```python
temperature = 35
if temperature > 30:
    print("It's a hot day.")
elif temperature > 20:
    print("It's a nice day.")
else:
    print("It's a cold day.")
```

Nested if-else Statement

```python
temp = 35
hum = 70
if temp > 30:
    if hum > 60:
        print("It's a hot and humid day.")
    else:
        print("It's a hot day.")
else:
    print("It's not a hot day.")
```

Ternary Conditional Statement

```python
temp = 35
message = "hot" if temp > 30 else "nice" if temp > 20 else "cold"
print(f"It's a {message} day.")
```

Match Case Statement

```python
color = "green"
match color:
    case "red":
        print("Red found")
    case "green":
        print("Green found")
    case "blue":
        print("Blue found")
    case _:
        print("No color found")
```
