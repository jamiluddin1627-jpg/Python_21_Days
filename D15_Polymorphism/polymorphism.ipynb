Method Overriding

```python
# multiple classes ----> multiple methods ----> same method name ----> different parameters ----> polymorphism

class Animal:
    def __init__(self, name):
        self.name = name

    def make_sound(self):
        pass

class Dog(Animal):
    def make_sound(self):
        return f"{self.name} barks: Woof! Woof!"

class Cat(Animal):
    def make_sound(self):
        return f"{self.name} meows: Meow! Meow!"

class Bird(Animal):
    def make_sound(self):
        return f"{self.name} chirps: Tweet! Tweet!"

# Demonstrating polymorphism
animals = [Dog("Buddy"), Cat("Whiskers"), Bird("Tweety")] 
# object of different classes stored in a list

for animal in animals:
    print(animal.make_sound())
```

**Duck Typing**

```python
class Dog:
    def sound (self):
        return "Woof! Woof!"

class Cat:
    def sound (self):
        return "Meow! Meow!"

class Bird:
    def sound (self):
        return "Tweet! Tweet!"

# function polymorphism
def animal_sound(animal):
    animal.sound() # calling the method of the object passed as argument
    return animal.sound() # returning the result of the method call

d = animal_sound(Dog()) # calling the function with a Dog object
c = animal_sound(Cat()) # calling the function with a Cat object
b = animal_sound(Bird()) # calling the function with a Bird object
print(d)
print(c)
print(b)
```

Abstraction

```python
from abc import ABC, abstractmethod

class Animal(ABC): # inheriting from ABC to create an abstract base class
    @abstractmethod # decorator to define an abstract method
    def sound(self):
        pass

class Dog(Animal):
    def __init__(self, name):
        self.name = name

    def sound(self):
        return f"{self.name} says Woof! Woof!"

class Cat(Animal):
    def __init__(self, name):
        self.name = name

    def sound(self):
        return f"{self.name} says Meow! Meow!"

class Bird(Animal):
    def __init__(self, name):
        self.name = name

    def sound(self):
        return f"{self.name} says Tweet! Tweet!"

d = Dog("Buddy")
c = Cat("Whiskers")
b = Bird("Tweety")
print(d.sound())
print(c.sound())
print(b.sound())
```
