To-Do List

```python
print("Welcome to To-Do List")
todo_list = []

# user interface
print("\nMenu")
print("1. Add Task")
print("2. View Task")
print("3. Remove Task")
print("4. Exit Task")

while True:
    choice = input("\nEnter your choice: (1-4)\n")
    
    # Adding Task
    if choice == '1':
        task = input("Enter your task: ")
        todo_list.append(task)
        print(f"'{task}' added successfully.")
        
    # View Task
    elif choice == '2':
        if len(todo_list) == 0:
            print("Your list is empty.")
        else:
            print("Your task list:")
            for index, task in enumerate(todo_list, start=1):
                print(f"{index}: {task}")
                
    # Removing Task
    elif choice == '3':
        if len(todo_list) == 0:
            print("Your list is empty.")
        else:
            print("Your current tasks:")
            for index, task in enumerate(todo_list, start=1):
                print(f"{index}: {task}")
            
            task_num = input("Which task do you want to remove? ")
            if task_num.isdigit():
                task_num = int(task_num)
                if 1 <= task_num <= len(todo_list):
                    removed_task = todo_list.pop(task_num - 1)
                    print(f"'{removed_task}' removed successfully.")
                else:
                    print("Invalid task number.")
            else:
                print("Please! Enter a valid number.")
                
    # Exit from to-do list
    elif choice == '4':
        print("Exit from the list manager.")
        break
        
    else:
        print("Invalid choice. Please enter a number between 1 and 4.")
```
