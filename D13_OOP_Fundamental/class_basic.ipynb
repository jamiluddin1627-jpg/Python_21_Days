Creating Class

```python
class Student:
    total_students = 0  # class attribute
    
    # Constructor: Initializes instance attributes and increments total_students
    def __init__(self, name, id):
        self.id = id  # instance attribute
        self.name = name  # instance attribute
        Student.total_students += 1

    # Instance method: Works with individual object data (uses 'self')
    def display_info(self):
        print(f"Name: {self.name}, ID: {self.id}")

    # -------------- CLASS METHOD --------------
    @classmethod
    def get_total_students(cls):
        return cls.total_students

    # -------------- STATIC METHODS --------------
    @staticmethod
    def institution_name():
        return "XYZ University"

    # Static method for ID validation
    @staticmethod
    def is_valid_id(student_id):
        # Checks if ID contains only digits and is non-empty
        return student_id.isdigit() and len(student_id) > 0

    # Static method for Name validation
    @staticmethod
    def is_valid_name(name):
        # Removes spaces and checks if the remaining string contains only alphabetic letters
        cleaned_name = name.replace(" ", "")
        return len(cleaned_name) > 0 and cleaned_name.isalpha()


# 1. List to store student objects
students_list = []

# 2. Get the number of students from user input
n = int(input("How many students' info do you want to enter? "))

# 3. Loop to take input, validate name and ID, and create Student objects
for i in range(n):
    print(f"\n--- Entering details for Student {i + 1} ---")

    # Loop until a valid Name is provided
    while True:
        name = input("Enter name (letters only): ")

        # Calling the static method for Name validation
        if Student.is_valid_name(name):
            break
        print("Invalid name! Name must contain alphabetic letters and spaces only. Please try again.")

    # Loop until a valid ID is provided
    while True:
        student_id = input("Enter ID (digits only): ")

        # Calling the static method for ID validation
        if Student.is_valid_id(student_id):
            break
        print("Invalid ID! ID must contain digits only. Please try again.")

    # Create a new Student object
    s = Student(name, student_id)

    # Store the object in the list
    students_list.append(s)
    #--------for loop end--------

# 4. Loop through the list to display all students' info
print("\n=========================")
print("All Students' Details:")
print("=========================")

for student in students_list:
    print(f"\n--- Student {students_list.index(student) + 1} Details ---")
    print(f"Institution: {student.institution_name()}")
    student.display_info()

# Display total student count using the class method
print("\nTotal number of students:", Student.get_total_students())
```
