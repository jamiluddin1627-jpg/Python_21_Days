**Tuple & Sets**

```python
tuples = (0, 1, 5, 6, 9, 10)
print(tuples[0])
print(tuples[-3:]) # Prints the last three elements of the tuple
print(tuples[-5]) # Prints the fifth element from the end of the tuple
```

**Tuples Operation and Function**

```python
tuple1 = (0, 1, 5, 6, 9, 10)
tuple2 = ('python', 'java', 'c++')
print(tuple1 + tuple2) # Concatenates two tuples
```

```python
print(len(tuple1))
print(max(tuple1))
print(min(tuple1))
```

```python
list1 = [1, 2, 3, 4, 5]
print(tuple(list1))
print(type(tuple(list1))) # Prints the type of the converted list to tuple
```

**Tuple Unpacking**

```python
tp = (2, 5, 6, 7, 8, 9)
a, b, c, d, e, f = tp
print(a, b, c, d, e, f)
```

**Sets**

Creating a set

```python
set1 = {1, 2, 3, 4, 5}
set2 = {4, 5, 6, 7, 8}
set3 = set1 & set2  # Intersection
set4 = set1 | set2  # Union
set5 = set1 - set2  # Difference
set6 = set1 ^ set2  # Symmetric Difference
print(set3)
print(set4)
print(set5)
print(set6)
```

```python
set1 = {1, 3, 2, 2, 4, 5}
set2 = {'a', 'c', 'd', 'e', 'b'}
print(set1)
print(set2)
```

```python
# List to Set
listToSet = set([4, 5, 6])
print(listToSet)
```

```python
# Accessing the value of set
st = {'a', 'b', 'c', 'd', 'e'}
for i in st:
    print(i, end = " ")
print("\n")
# Accessing the value of set using set comprehension
st = {i for i in 'abcde'}
print(st)
```
