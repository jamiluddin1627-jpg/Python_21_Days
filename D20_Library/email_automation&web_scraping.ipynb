```python
import smtplib

# take data from user
# sender mail , reciever mail, sub and message

s_mail = input("Sender Mail: ")
r_mail = input("Reciever Mail: ")
subject = input("Sub: ")
message = input("Message: ")

# Formatting subject and message
#msg = "Subject: {}\n\n".format(subject, message)

msg = f"Subject: {subject}\n\n{message}"

# Server Setup

server = smtplib.SMTP("smtp.gmail.com", 587)
server.starttls()

# Server Login
server.login(s_mail, "dugd gxgq jmlw xszi")

# Send Mail
server.sendmail(s_mail, r_mail, msg)
server.quit()
```

```python
import requests
from bs4 import BeautifulSoup
```

```python
# Web page Fetch
url = "https://webscraper.io/"
response = requests.get(url)
print(response.status_code)

print(response.text[:1000])
```

Parse HTML with BeautifulSoup + LXML

```python
soup = BeautifulSoup(response.text, "lxml")
```

```python
soup.title.string
```

```python
h2 = soup.find_all("h2")
for h in h2:
    print(h.text.strip())
```
