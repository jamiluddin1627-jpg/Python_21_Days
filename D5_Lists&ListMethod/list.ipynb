**List: **  ***Add List Item***

```python
my_list = [1, 2, 3]
print
my_list.append(4)
print(my_list)  # Output: [1, 2, 3, 4]

my_list.insert(1, 'a') #(index, element) Insert 'a' at index 1
print(my_list)  # Output: [1, 'a', 2, 3, 4] inserted 'a' at index 1

my_list.extend([5, 6])
print(my_list)  # Output: [1, 2, 3, 4, 5, 6] extended the list with another list
```

***Remove List Item***

```python
fruits = ['apple', 'banana', 'cherry', 'date']
fruits.remove('banana')
print(fruits)  # Output: ['apple', 'cherry', 'date']
fruits.pop(1)
print(fruits)  # Output: ['apple', 'date']
fruits.clear() # Remove all elements from the list
print(fruits)  # Output: []
```

***List Comprehension***

```python
squares =[i**i for i in range(6) if i%2==0]
print(squares)
```

```python
fruits = ['apple', 'banana', 'cherry', 'date']
# Using a for loop to iterate over the list
uppercase_fruits = [i.upper() for i in fruits]
print(uppercase_fruits)  # Output: ['APPLE', 'BANANA', 'CHERRY', 'DATE']
```
