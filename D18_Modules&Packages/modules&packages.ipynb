```python
# Math Module
import math
math.sqrt(64)
math.pow(2, 4)
math.pi
math.factorial(5)
math.log10(10)
```

```python
# Random 
import random

random.randint(1,50)

random.sample(range(1, 50), 10) # 10 samples range from 1 to 50

random.random()

for i in range(5):
    print(random.random(), end = "\n")
    
for i in range(10):
    print(random.randint(1, 50), end = "\n")
    
fruits = ["apple", "banana", "cherry", "graps"]
print(random.choice(fruits))
```
