**Scope of Variable:** A variable's scope refers to the region of the code where the variable is accessible. In Python, there are four types of variable scopes:

Local Variable

```python
#local variable
#-----block-----
def my_function():
    local_var = 10
    print(local_var)
my_function() # This will print 10 and its the end of the block
# print(local_var)  # This will raise a NameError since 'local_var' is not accessible outside the function
```

Global Variable

```python
#Global variable
global_var = 100
#-----block-----
def my_function():
    print(global_var) # This will print 100 since 'global_var' is accessible inside the function
my_function() # This will print 100 and its the end of the block
print(global_var) # This will also print 100 since 'global_var' is a global variable
```

**User Input and Output**

```python
f_name = input("Enter your first name: ")
l_name = input("Enter your last name: ")
full_name = f"{f_name} {l_name}"
print("Your full name is:", full_name)
```

Split Function

```python
fname, lname = input("Enter your first and last name separated by a space: ").split()
fullname = f"{fname} {lname}"
print("Your full name is:", fullname)
```
