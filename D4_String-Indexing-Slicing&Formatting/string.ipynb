**String**

```python
a = "bangladesh"
#String slicing
print(a[0:5:1]) 
print(a[:-4])
print(a[-11:-4])
```

```python
#Indexing and negative indexing
a = "bangladesh"
print(a[9])  # Positive indexing
print(a[-1]) # Negative indexing
```

**String-Method**

```python
s = "goBeshona leaRning"
print(s.upper())
print(s.lower())
print(s.title())
print(s.capitalize())
print(s.swapcase())
```

**Strip**

```python
a = "  Gobeshona" 
print(a)
print(a.strip())
```

**Replace**

```python
s = "Gobeshona Learning"
print(s)
print(s.replace("Learning", "Training"))
```

**Split**

```python
s = "Gobeshona Learning"
print(s.split("a"))
```
