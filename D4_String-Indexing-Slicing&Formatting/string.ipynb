**String**

```python
a = "bangladesh"
print(type(a))
#String slicing
print(a[0:5:1]) 
print(a[:-4])
print(a[-11:-4])

s = "goBeshona leaRning"
x = " and Analytics"
print(s + x)
```

```python
#Indexing and negative indexing
a = "bangladesh"
print(a[9])  # Positive indexing
print(a[-1]) # Negative indexing
```

**String-Method**

```python
s = "goBeshona leaRning"
a = " and Analytics"
print(s.upper())
print(s.lower())
print(s.title())
print(s.capitalize())
print(s.swapcase())
```

**Strip**

```python
a = "  Gobeshona" 
print(a)
print(a.strip())
```

**Replace**

```python
s = "Gobeshona Learning"
print(s)
print(s.replace("Learning", "Training"))
```

**Split**

```python
s = "Gobeshona Learning"
print(s.split("a"))
```

**Length & Count**

```python
s = "Gobeshona Learning"
print(len(s))
print(s.count("o"))
```

**String Formatting**

```python
name = input("Enter your name: ")
age = int(input("Enter your age: "))

print(f"My name is {name} and I am {age} years old.")
```
