for loop

```python
for i in range(5):
    print(i)
```

for loop with list

```python
list = ["apple", "banana", "cherry"]
for fruit in list:
    print(fruit)
```

for loop with dictionary

```python
dict = { "name" : "Jamil", "age" : "26", "result" : 3.90}

for info in dict:
    print(info, ":", dict[info])
```

for-else statement

```python
for i in range (5):
    if i == 3:
        break
    print(i)
else:
    print("execute properly")
```

```python
for i in range (5):
    if i == 3:
        continue # skip i == 3 iteration
    print(i)
else:
    print("execute properly")
```

nested for loop

```python
list1 = ["apple", "banana"]
list2 = ["red", "green"]
for i in list1:
    for j in list2:
        print(i, j)
```

while loop

```python
a = 0
while(a<5):
    a += 2
    print("hello")
```
