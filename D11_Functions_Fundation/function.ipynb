Creating Function and Function Calling

```python
def say_hello():
    print("Hello, world!")
say_hello()
```

Argument Parameter

```python
def greet(name):
    print(f"Hello, {name}!")
greet("Jamil")
```

Multiple Arguments Parameter

```python
def about(name, age):
    print(f" I am {name} and I am {age} years old.")
about("Jamil", 25)
```

Arbitrary Argument

```python
def info(*args):
    for i in args:
        print(i)
info("Jamil", 25, "Python Developer")
```

Keyword Arguments

```python
def about(name, age, GPA):
    print(age, name, GPA)
about( name = "Jamil", age = 25, GPA = 3.9)
```

Arbitrary Keyword

```python
def about(**args):
    for key, value in args.items():
        print(f"{key}: {value}")
about(name="Jamil", age=25, GPA=3.9)
```

Default Parameter

```python
def country(name = "Bangladesh"):
    print(f"I am from {name}.")
country()
```

Return

```python
def sum(a, b):
    return a + b
print(sum(5, 10))
```

```python
def sub(a, b):
    return a - b
subtraction = sub(10, 5)
print(subtraction)
```

Print Vs Return

```python
def sum(a, b):
    print(a+b)
value = sum(5, 10)
print(value)  # This will print None because the function does not return a value
```
