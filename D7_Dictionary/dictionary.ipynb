**Key Characteristics of Dictionaries**

Key–Value Mapping: Each key is unique and maps to a value.

Unordered: Items don’t maintain insertion order in older versions (<3.7), but from Python 3.7+, dictionaries preserve insertion order.

Mutable: You can add, remove, or update key–value pairs.

Keys Must Be Hashable: Immutable types like strings, numbers, and tuples can be keys. Lists or other dictionaries cannot.

Fast Lookup: Optimized for quick access by key.

```python
# dictionary 
student = {
    "name" : "Jamil",
    "age" : 20,
    (80, 81, 79) : "marks"
    
}
print(student)
print(type(student))
```

```python
# using dict()
students = dict(name = "Alice", age = 20, marks = (80, 78, 50))
print(students)
```

**Accessing Dictionary**

```python
student = {
  "name": "Karim",
  "age": 20,
  (80, 78, 50): "marks"
}

print(student["name"])
```

**Modifying a dictionary**

```python
# Modifying a dictionary
student = {
    "name": "Jamil",
    "age": 20,
    "dept" : "CSE"
}
student["age"]  = 26
student["name"] = "Md. Jamil Uddin"
print(student)
```

**Iterating Through Dictionary**

```python
student = {
    'name': 'Jamil',
    'age': 20,
    'dept': 'CSE'
}
for i, j in student.items(): # items() method returns a view object that displays a list of a dictionary's key-value tuple pairs.
    print(i, ":", j)
```

**Nested Dictionary**

```python
student_info = {
    'student1' : {'name' : 'Jamil', 'age' : 26, 'dept' : 'CSE'},
    'student2' : {'name' : 'Karim', 'age' : 22, 'dept' : 'EEE'}
}
print(student_info)
```

```python
#Accessing a nested dictionary
print(student_info['student1']['name'])
#Modifying a nested dictionary
student_info['student1']['age'] = 27
print(student_info['student1']['age'])
```

**Iteration through Nested dictionary**

```python
for outer_key, outer_value in student_info.items():
    print(outer_key)
    for inner_key, inner_value in outer_value.items():
        print(inner_key, ":", inner_value)
```
