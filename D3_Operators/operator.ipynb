**Operator**

```python
num1 = int(input("Enter the first number: "))
num2 = int(input("Enter the second number: "))
add = num1 + num2
sub = num1 - num2
mul = num1 * num2
div = num1 / num2
exp = num1 ** num2
f_div = num1 // num2
rem = num1 % num2
print("Addition: ", add)
print("Subtraction: ", sub)
print("Multiplication: ", mul)
print("Division: ", div)
print("Exponentiation: ", exp)
print("Floor Division: ", f_div)
print("Remainder: ", rem)
```

**Logical Operator**

```python
a = 10
b = 25
print(a>b or b>a)
print(a>b and b>a)
print(not(a>b))
```

Membership Operator

```python
a = 5
b = [1, 2, 3, 4, 5]
print(a in b)
```

Identity Operator

```python
a = 10
b = a
c = 100

print(a is b)

print(a is c)
```
